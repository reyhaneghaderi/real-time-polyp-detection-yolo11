```python
from ultralytics import YOLO
import os, shutil

DATA_YAML = "/content/Bsc:-Train_PSC1toC5-7/data.yaml"
VAL_IMGS  = "/content/Bsc:-Train_PSC1toC5-7/valid/images"

best = YOLO("/content/runs/detect/train2/weights/best.pt")

# This computes mAP, Precision, Recall, and saves plots (PR curve, confusion matrix)
val_metrics = best.val(
    data=DATA_YAML,
    split="val",      # use validation set only
    imgsz=640,
    batch=32,
    conf=0.001,       # low conf for metric computation
    iou=0.5,
    plots=True,
    save_json=True
)

# --- Print key numbers (works for many Ultralytics versions) ---
try:
    print(f"mAP50:     {val_metrics.box.map50:.4f}")
    print(f"mAP50-95:  {val_metrics.box.map:.4f}")
    if hasattr(val_metrics.box, "mp"):
        print(f"Precision: {val_metrics.box.mp:.4f}")
    if hasattr(val_metrics.box, "mr"):
        print(f"Recall:    {val_metrics.box.mr:.4f}")
except Exception as e:
    print("Metrics object structure changed; inspect val_metrics:", e)
    print(type(val_metrics), dir(val_metrics))
```

```python
# --- 4) (Optional) Save visual predictions on the validation set for a quick look ---
best.predict(
    source=VAL_IMGS,
    conf=0.25,    # raise/lower if you want fewer/more boxes
    iou=0.5,
    save=True,    # saves annotated images
    save_txt=True # saves YOLO txt labels per image
)

print("Validation artifacts:")
print(" - Metrics & plots: /content/runs/detect/val*/")
print(" - Annotated preds: /content/runs/detect/predict*/")
```
